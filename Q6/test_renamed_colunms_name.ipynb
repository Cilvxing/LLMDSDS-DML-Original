```python
import gzip
import json
import ast

file_path = r"D:\Adata\P5_data\data\beauty\meta.json.gz"

def try_parse(line):
    """尝试多种方式解析一行"""
    line = line.strip()
    if not line:
        return None
    
    # 1. 标准JSON
    try:
        return json.loads(line)
    except:
        pass
    
    # 2. 单引号JSON (Python dict字符串)
    try:
        return ast.literal_eval(line)
    except:
        pass
    
    # 3. 去掉可能的BOM后再试
    try:
        return json.loads(line.lstrip('\ufeff'))
    except:
        pass
    
    return None

with gzip.open(file_path, 'rt', encoding='utf-8') as f:
    for i, line in enumerate(f):
        if i >= 3:
            break
        print(f"\n=== 第{i+1}行原始内容 ({len(line)}字符) ===")
        print(line[:500])
        
        result = try_parse(line)
        if result:
            print(f"\n✅ 解析成功! 字段: {list(result.keys()) if isinstance(result, dict) else '非字典类型'}")
            print(f"预览: {json.dumps(result, indent=2, ensure_ascii=False)[:800]}")
        else:
            print("❌ 解析失败，请检查原始内容格式")
```

```python
import gzip
import json
from tqdm import tqdm
from pathlib import Path

input_file = "./raw_data/meta_Beauty_and_Personal_Care_Extracted.jsonl.gz"
output_file = str(Path(input_file).with_suffix('').with_suffix('')) + "_renamed1.jsonl.gz"

# 先计算总行数
print("正在统计文件行数...")
with gzip.open(input_file, 'rt', encoding='utf-8') as f:
    total_lines = sum(1 for _ in f)

print(f"总行数: {total_lines:,}，开始处理...")

with gzip.open(input_file, 'rt', encoding='utf-8') as fin, \
     gzip.open(output_file, 'wt', encoding='utf-8') as fout:
    
    for line in tqdm(fin, total=total_lines, desc="重命名字段"):
        data = json.loads(line)
        
        # 将 description 改为 newdescription
        if 'description' in data:
            data['newdescription'] = data.pop('description')
        
        fout.write(json.dumps(data, ensure_ascii=False) + '\n')

print(f"✅ 完成！文件保存至: {output_file}")
```

```python
import gzip
import json
from tqdm import tqdm
from pathlib import Path

input_file = "./raw_data/meta_Beauty_and_Personal_Care_Extracted_renamed1.jsonl.gz"
output_file = str(Path(input_file).with_suffix('').with_suffix('')) + "_renamed2.jsonl.gz"

# 先计算总行数
print("正在统计文件行数...")
with gzip.open(input_file, 'rt', encoding='utf-8') as f:
    total_lines = sum(1 for _ in f)

print(f"总行数: {total_lines:,}，开始处理...")

with gzip.open(input_file, 'rt', encoding='utf-8') as fin, \
     gzip.open(output_file, 'wt', encoding='utf-8') as fout:
    
    for line in tqdm(fin, total=total_lines, desc="重命名字段"):
        data = json.loads(line)
        
        # 将 parent_asin 改为 asin
        if 'parent_asin' in data:
            data['asin'] = data.pop('parent_asin')

        # 将 features 改为 description
        if 'features' in data:
            data['description'] = data.pop('features')
            
        fout.write(json.dumps(data, ensure_ascii=False) + '\n')

print(f"✅ 完成！文件保存至: {output_file}")
```

```python
# 我现在拥有的新版本的评论数据集Beauty_and_Personal_Care.jsonl,
# 修改字段名字，user_id变为reviewerID，text变为reviewText，rating变为overall，timestamp变为unixReviewTime，其余字段不变。

import json
from pathlib import Path
from tqdm import tqdm

input_file = "./raw_data/Beauty_and_Personal_Care.jsonl"
output_file = str(Path(input_file).with_suffix('').with_suffix('')) + "_renamed1.jsonl"
# input_file = "./raw_data/Beauty_and_Personal_Care_Aligned.jsonl"

# 先统计总行数
print("统计文件行数...")
total_lines = sum(1 for _ in open(input_file, 'r', encoding='utf-8'))

print(f"总行数: {total_lines:,}，开始转换字段...")

with open(input_file, 'r', encoding='utf-8') as fin, \
     open(output_file, 'w', encoding='utf-8') as fout:
    
    for line in tqdm(fin, total=total_lines, desc="转换字段"):
        data = json.loads(line)
        
        # 执行字段重命名
        field_mappings = {
            'user_id': 'reviewerID',
            'text': 'reviewText', 
            'rating': 'overall',
            'title':'summary',
            'timestamp': 'unixReviewTime',
            'asin':'Secondary-asin',
        }
        
        for old_name, new_name in field_mappings.items():
            if old_name in data:
                data[new_name] = data.pop(old_name)
        
        fout.write(json.dumps(data, ensure_ascii=False) + '\n')

print(f"✅ 字段重命名完成！")
print(f"输出文件: {output_file}")
```
